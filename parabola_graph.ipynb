# Parabola Graph Visualization
This notebook displays a graph of a parabola using the quadratic function y = x²

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Generate x values
x = np.linspace(-10, 10, 100)

# Calculate y values for the parabola y = x^2
y = x**2
```

```python
# Create the plot
plt.figure(figsize=(10, 6))
plt.plot(x, y, 'b-', linewidth=2, label='y = x²')
plt.grid(True, alpha=0.3)
plt.xlabel('x', fontsize=12)
plt.ylabel('y', fontsize=12)
plt.title('Parabola Graph: y = x²', fontsize=14, fontweight='bold')
plt.axhline(y=0, color='k', linewidth=0.5)
plt.axvline(x=0, color='k', linewidth=0.5)
plt.legend(fontsize=12)
plt.tight_layout()
plt.show()
```
